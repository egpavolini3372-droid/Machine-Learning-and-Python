```python
vector = np.array([3,-4,0])

norma_l1 = np.linalg.norm(vector, ord = 1) # suma los valores absolutos del vector 3 + 4 = 7
norma_l2 = np.linalg.norm(vector, ord = 2) # la raiz cuadrada de la suma de los cuadrados de los componentes (9+16+0)^0.5 = 5
norma_inf = np.linalg.norm(vector, ord = np.inf) # El valor absoluto del elemento mas grande
norma_cero= np.linalg.norm(vector, ord = 0) # los valores diferentes de cero

print(round(norma_l1,2))
print(round(norma_l2,2))
print(round(norma_inf,2))
print(round(norma_cero,2))
```

```python
#CALIFICACION DE PELICULAS DE ACUERDO A SI ES COMEDIO (PRIMERA POSICION) O SI ES ROMANCE(SEGUNDA POSICION)

peliculas = np.array([[5,2],[1,5],[4,4]])

# Matriz de peso que predice genero d ela pelicula: (posicion 1) accion, (posicion 2) comedia, (posicion 3) familia
pesos_general = np.array([[1.1, 0.4, 0.1],[0.2,0.5,1.2]])

print(peliculas.shape)
print(pesos_general.shape)

```

```python
scores_prediccion = peliculas @ pesos_general
print(scores_prediccion)
```

```python
# calificacion de peliculas de acuerdo a gusto de usuario por accion, comedia, drama, ciencia ficcion

usuarios = np.array([[0.9, 0.2, 0.1, 0.8],[0.1, 0.8, 0.7, 0.2],[0.8, 0.7, 0.1, 0.1]])
# calificacion de peliculas: Blade Runner, Super cool
peliculas = np.array([[0.8, 0.1, 0.1, 0.9],[0.2, 0.8, 0.9, 0.1]])

print(usuarios.shape)
print(peliculas.shape)
```
