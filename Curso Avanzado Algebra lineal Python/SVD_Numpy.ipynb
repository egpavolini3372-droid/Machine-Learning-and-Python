```python
import numpy as np

A = np.array([[1,2],[3,4],[5,6]])
print(A.shape)
```

```python
U, S, VT = np.linalg.svd(A) # Desempaquetar las matrices de rotación y escalamiento
print(f"U: {U}") #
print(f"S: {S}") # Vector con los valores de la diagonal
print(f"VT: {VT}") # Sus columnas son los eigevectores de la matriz A^T.A
```

```python
Sigma = np.zeros(A.shape)

Sigma[:len(S), :len(S)] = np.diag(S)
print(Sigma)
```

```python
A_reconstruida = U @ Sigma @ VT
print(A_reconstruida)
```

```python
# SISTEMA DE CALIFICACION DE 0.0 A 5.0 PARA TRES PELICULAS (DUNE, SHREK, LOS CASA FANTASMAS)
patron_scifi = np.array([5,1,4]) # calificacion de tres peliculas de usuario con gusto de ciencia ficcion
patron_comedia = np.array([1,5,3]) # calificacion de tres peliculas de usuario con gusto de comedia

A_signal = np.array([patron_scifi, patron_comedia, patron_scifi *0.5 + patron_comedia*0.5, patron_scifi *0.8 + patron_comedia*0.2])

print(A_signal)
```

```python
ruido = np.random.randn(4,3)*0.3 # agregar ruido a la señal

A_ruidosa = A_signal + ruido
print(A_ruidosa)
```

```python
U, S, VT = np.linalg.svd(A_ruidosa)
print(np.round(S,2))
```

```python
k = 3

U_k = U[:, :k]
S_k = np.diag(S[:k])
VT_k = VT[:k, :]

A_filtrada = U_k @ S_k @ VT_k
print(np.round(A_filtrada,2))
```
