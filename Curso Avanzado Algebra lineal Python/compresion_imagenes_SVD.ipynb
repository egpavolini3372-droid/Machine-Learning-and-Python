```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from skimage.transform import resize
from skimage.color import rgb2gray

imagen_original = rgb2gray(data.rocket())
imagen = resize(imagen_original, (300, 200))
```

```python
U, S, VT = np.linalg.svd(imagen)
print(VT.shape)
```

```python
def recontruir_con_k(k):
  U_k = U[:, :k]
  S_k = np.diag(S[:k])
  VT_k = VT[:k, :]
  imagen_reconstruida = np.dot(U_k, np.dot(S_k, VT_k))
  return imagen_reconstruida
```

```python
k_valores = [5,20,50]
imagenes_recontruidas = [recontruir_con_k(k) for k in k_valores]
```

```python
fig, axes = plt.subplots(1, 4, figsize = (20,5))

axes[0].imshow(imagen, cmap ='gray')
axes[0].set_title('Imagen original')

for i, k in enumerate(k_valores):
  axes[i+1].imshow(imagenes_recontruidas[i], cmap ='gray')
  axes[i+1].set_title(f'reconstruccion con k ={k} ')

plt.show()
```
