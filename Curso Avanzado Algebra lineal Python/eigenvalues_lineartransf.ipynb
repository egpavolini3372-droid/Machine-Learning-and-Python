```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import pandas_datareader as web
import datetime as dt
```

```python
v_ori = np.array([2,1])

mat_rot = np.array([[0, -1],[1, 0]])
mat_esc = np.array([[2,0],[0,0.5]])
mat_inc = np.array([[1,1],[0,1]])

v_rot = np.dot(mat_rot,v_ori)
v_inc = np.dot(mat_inc,v_ori)
v_esc = np.dot(mat_esc,v_ori)

print(v_rot)
print(v_inc)
print(v_esc)
```

```python
mat_comp = mat_inc @ mat_esc @ mat_rot
print(mat_comp)
#
v_comp = np.dot(mat_comp,v_ori)
print(v_comp)
```

```python
fig, (ax1,ax2,ax3,ax4) = plt.subplots(1,4,figsize=(24,8))
fig.suptitle('tipos de transformaciones lineales aplicadas a un vector')
def graficar_transformaciones(ax, title,v_transformado):
  ax.plot([0,v_ori[0]],[0,v_ori[1]],'--',color ='gray',label ='vector original')
  ax.quiver(0,0,v_transformado[0],v_transformado[1], angles ='xy', scale_units ='xy',scale =1, color ='red', label ='vector transformado')
  ax.set_title(title)
  ax.grid(True)
  ax.set_xlim(-3,4)
  ax.set_ylim(-3,4)
  ax.legend()

graficar_transformaciones(ax1,'rotacion',v_rot)
graficar_transformaciones(ax2,'escala',v_esc)
graficar_transformaciones(ax3,'inclinacion',v_inc)
graficar_transformaciones(ax4,'compuesta',v_comp)
plt.show()
```

```python
# Calculo eigenvalues y eigenvectors

Matriz_A = np.array([[3,1],[1,2]])

eingenvalues, eigenvectors = np.linalg.eig(Matriz_A)

print(f'Autovalores: {eingenvalues}')
print(f'Autovectores: {eigenvectors}')
```

```python
# A.v = lambda * v

# se extraen los dos autovectores de la matriz de autovectores
v1 = eigenvectors[:,0]
v2 = eigenvectors[:,1]

# se extraen los autovalores del autovector

lambda1 = eingenvalues[0]
lambda2 = eingenvalues[1]

# se calcula uno de los lados de la ecuacion
A_v1 = Matriz_A @ v1
lambda_v1 = lambda1 * v1
print(f"lado izquierdo de la ecuacion: {A_v1} ")
print(f'lado derecho de la ecuacion: {lambda_v1}')

plt.figure(figsize = (8,8))
plt.quiver(0,0,v1[0],v1[1],angles ='xy', scale_units ='xy', scale = 1, color = 'blue', label = 'eigenvector original')
plt.quiver(0,0,A_v1[0],A_v1[1],angles ='xy', scale_units ='xy', scale = 1, color = 'red', label = 'eigenvector escalado')

plt.xlim(0,4); plt.ylim(0,4)
plt.legend()
plt.title('eigenvector original vs escalado')
plt.grid(True)
plt.show()
```
