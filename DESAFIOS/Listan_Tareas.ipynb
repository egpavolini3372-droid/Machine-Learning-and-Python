```python
tareas = {}

while True:
  print("\n1. Agregar  2. Ver  3. Completar  4. Salir")
  opcion = input("Elige: ")

  if opcion == "1":
        tarea = input("Nombre de la tarea: ")
        tareas[tarea] = {"completada": False}
        print("✅ Tarea agregada!")

  elif opcion == "2":
        if len(tareas) == 0:
            print("No hay tareas aún.")
        else:
           for nombre, detalle in tareas.items():
                estado = "✅" if detalle["completada"] else "[ ]"
                print(f"{estado} {nombre}")

  elif opcion == "3":
        if not tareas: # Check if dictionary is empty
            print("No hay tareas para completar.")
        else:
            print("Tareas actuales:")
            indexed_tareas = {}
            for i, (nombre, detalle) in enumerate(tareas.items()):
                estado = "✅" if detalle["completada"] else "[ ]"
                print(f"{i + 1}. {estado} {nombre}") # Print enumerated list
                indexed_tareas[str(i + 1)] = nombre # Store mapping: number (as string) -> task name

            seleccion_str = input("Ingrese el NÚMERO de la tarea a completar: ")

            if seleccion_str in indexed_tareas:
                tarea_a_completar = indexed_tareas[seleccion_str] # Get task name using the number
                respuesta = input(f"¿La tarea '{tarea_a_completar}' ha sido completada? (si/no): ").lower()
                if respuesta == "si":
                    tareas[tarea_a_completar]["completada"] = True
                    print(f"✅ Tarea '{tarea_a_completar}' marcada como completada.")
                elif respuesta == "no":
                    tareas[tarea_a_completar]["completada"] = False
                    print(f"🔄 Tarea '{tarea_a_completar}' marcada como pendiente.")
                else:
                    print("Respuesta inválida. Use 'si' o 'no'.")
            else:
                print("❌ Número de tarea inválido o la tarea no existe.")

  elif opcion == "4":
        print("👋 ¡Hasta luego!")
        break
```
