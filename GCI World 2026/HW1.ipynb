```python
# Implement a function that takes a 1D NumPy array of integers and returns an array containing elements that are multiples of 5 and
# leave a remainder of 1 when divided by 2

def multiplies_5(array):
  # Use boolean indexing to filter the array directly
  # (array % 2 == 1) checks for odd numbers
  # (array % 5 == 0) checks for multiples of 5
  filtered_array = array[(array % 2 == 1) & (array % 5 == 0)]
  return filtered_array

a = np.array([1, 5, 10, 3, 4, 25, 30,35])
print(multiplies_5(a))
```
