```python

class DivisionError(Exception):
  """Error en operacion"""
  pass

a = 0
b = 0

try:
  a = int(input("ingresa un numero: "))
  b = int(input("ingresa otro numero: "))
  if b == 2:
    raise DivisionError("No se puede dividir por 2")
  resultado = a / b
  print(resultado) # Moved print(resultado) inside the try block
except ValueError as e:
  print(f"el valor que digito no es un numero valido, error tipo: {e}")
except ZeroDivisionError as e:
  print(f"No es posible dividir por cero, error tipo {e}")
finally:
  print("Print desde Finally")
```
