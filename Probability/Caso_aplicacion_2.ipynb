```python
from scipy.stats import norm

alpha = 0.975
# Calcula el cuantil (Z-score)
z_score = norm.ppf(alpha)

print(z_score)
# Resultado aproximado:
```

```python
import math
n=289
sr=math.sqrt(n)
mean = data["Duración promedio llamadas al mismo operador"].mean()
std_dev = data["Duración promedio llamadas al mismo operador"].std()

IC_1=mean-(z_score*std_dev/sr)
IC_2=mean +(z_score*std_dev/sr)
print(IC_1, IC_2)
```

```python
# limite inferior IC

# Define the degrees of freedom (df) and percentile
df1 = n-1  # Example: Degrees of freedom
percentile1 = 1-0.025  # Example: 99th percentile (area to the left)

# Calculate the chi-square value (critical value)
chi_square_value1 = stats.chi2.ppf(q=percentile1, df=df1)

print(f"The chi-square value for {df1} degrees of freedom and the {percentile1*100}th percentile is: {chi_square_value1}")

# limite Superior IC

# Define the degrees of freedom (df) and percentile
df2 = n  # Example: Degrees of freedom
percentile2 = 0.025  # Example: 95th percentile (area to the left)

# Calculate the chi-square value (critical value)
chi_square_value2 = stats.chi2.ppf(q=percentile2, df=df2)

print(f"The chi-square value for {df2} degrees of freedom and the {percentile2*100}th percentile is: {chi_square_value2}")
```

```python
variance2 = data["Duración promedio llamadas a otros operadores"].var()

IC_1 = ((n-1)*variance2)/chi_square_value1
IC_2 = ((n-1)*variance2)/chi_square_value2
print(IC_1,IC_2)
```

```python
# Define your parameters:
probability = 0.95  # The cumulative probability (e.g., for a 95% confidence level)
dfn = 289             # Numerator degrees of freedom (df1)
dfd = 289            # Denominator degrees of freedom (df2)

# Calculate the F value (critical value)
f_value_1 = stats.f.ppf(q=probability, dfn=dfn, dfd=dfd)

probability2 = 0.05  # The cumulative probability (e.g., for a 95% confidence level)
dfn2 = 289             # Numerator degrees of freedom (df1)
dfd2 = 289            # Denominator degrees of freedom (df2)

# Calculate the F value (critical value)
f_value_2 = stats.f.ppf(q=probability2, dfn=dfn2, dfd=dfd2)


print(f"The F value for a cumulative probability of {probability} is: {f_value_1}")
print(f"The F value for a cumulative probability of {probability2} is: {f_value_2}")
```

```python
IC_2f = variance3/variance4*f_value_1
IC_1f = variance3/variance4*f_value_2
print(IC_1f,IC_2f)
```
