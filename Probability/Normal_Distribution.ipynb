```python
data =  np.array([98,102,104,98,100,102,96])

# Calculate basic statistics
mean_val = np.mean(data)
median_val = np.median(data)
std_dev = np.std(data) # Standard deviation
variance = np.var(data) # Variance
min_val = np.min(data) # Minimum value
max_val = np.max(data) # Maximum value
n_count = len(data)

print(f"Mean: {mean_val}")
print(f"Median: {median_val}")
print(f"Standard Deviation: {std_dev}")
print(f"Variance: {variance}")
print(f"Minimum: {min_val}")
print(f"Maximum: {max_val}")
print(f"n: {n_count}")
```

```python
# t studen Distribition
# Degrees of freedom
df = 6
# Significance level (alpha)
alpha = 0.975

# For a two-tailed test (e.g., 95% confidence):
# Probability q is 1 - alpha/2
q_two_tailed = 1 - alpha / 2
critical_t_two_tailed = stats.t.ppf(q_two_tailed, df)

print(f"Two-tailed critical t-value: {critical_t_two_tailed:.3f}")

# For a one-tailed test (e.g., right-tailed, alpha=0.05):
# Probability q is 1 - alpha
q_right_tailed = 1 - alpha
critical_t_right_tailed = stats.t.ppf(q_right_tailed, df)

print(f"Right-tailed critical t-value: {critical_t_right_tailed:.3f}")
```

```python
import math
n=7
sr=math.sqrt(n)

IC_1=mean_val -(-critical_t_right_tailed*std_dev/sr)
IC_2=mean_val +(-critical_t_right_tailed*std_dev/sr)
print(IC_1, IC_2)
```

```python
data2 =  np.array([3.01,3.05,2.99,2.99,3.0,3.02,2.98,2.99,2.97,3.02,3.01])

# Calculate basic statistics
mean_val2 = np.mean(data2)
median_val2 = np.median(data2)
std_dev2 = np.std(data2) # Standard deviation
variance2 = np.var(data2) # Variance
min_val2 = np.min(data2) # Minimum value
max_val2 = np.max(data2) # Maximum value
n_count2 = len(data2)

print(f"Mean: {mean_val2}")
print(f"Median: {median_val2}")
print(f"Standard Deviation: {std_dev2}")
print(f"Variance: {variance2}")
print(f"Minimum: {min_val2}")
print(f"Maximum: {max_val2}")
print(f"n: {n_count2}")
```

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# limite inferior IC

# Define the degrees of freedom (df) and percentile
df1 = n_count2  # Example: Degrees of freedom
percentile1 = 1-0.005  # Example: 99th percentile (area to the left)

# Calculate the chi-square value (critical value)
chi_square_value1 = stats.chi2.ppf(q=percentile1, df=df1)

print(f"The chi-square value for {df1} degrees of freedom and the {percentile1*100}th percentile is: {chi_square_value1}")

# limite Superior IC

# Define the degrees of freedom (df) and percentile
df2 = n_count2  # Example: Degrees of freedom
percentile2 = 0.005  # Example: 95th percentile (area to the left)

# Calculate the chi-square value (critical value)
chi_square_value2 = stats.chi2.ppf(q=percentile2, df=df2)

print(f"The chi-square value for {df2} degrees of freedom and the {percentile2*100}th percentile is: {chi_square_value2}")
```

```python
IC_1 = ((n_count2-1)*variance2)/chi_square_value1
IC_2 = ((n_count2-1)*variance2)/chi_square_value2
print(IC_1,IC_2)
```
