```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Define your parameters:
probability = 0.05  # The cumulative probability (e.g., for a 95% confidence level)
dfn = 3             # Numerator degrees of freedom (df1)
dfd = 50            # Denominator degrees of freedom (df2)

# Calculate the F value (critical value)
f_value = stats.f.ppf(q=probability, dfn=dfn, dfd=dfd)

print(f"The F value for a cumulative probability of {probability} is: {f_value}")
```
