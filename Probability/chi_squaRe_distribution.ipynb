```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Define the degrees of freedom (df) and percentile
df = 6  # Example: Degrees of freedom
percentile = 0.025  # Example: 95th percentile (area to the left)

# Calculate the chi-square value (critical value)
chi_square_value = stats.chi2.ppf(q=percentile, df=df)

print(f"The chi-square value for {df} degrees of freedom and the {percentile*100}th percentile is: {chi_square_value}")
```
