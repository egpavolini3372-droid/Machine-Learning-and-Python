```python
# Punto 3

mean_engineer = 22.74
mean_BM = 27.23
std_engineer = 2.99
std_BM = 3.06
n1=10
n2 =14

# eL ESTADISTICO DE PRUEBA PARA UNA HIPOTESIS DE LAS VARIANZAS POBLACIONALES ES: Sx^2/Sy^2

EP = (std_engineer**2)/(std_BM**2)
print(f" El estadistico de prueba es {EP}")

# La region de rechazo es F _(nx-1,ny-1)

# Define your parameters:
probability = 0.9  # The cumulative probability (e.g., for a 95% confidence level)
dfn = n1-1             # Numerator degrees of freedom (df1)
dfd = n2-1            # Denominator degrees of freedom (df2)

# Calculate the F value (critical value)
f_value_1 = stats.f.ppf(q=probability, dfn=dfn, dfd=dfd)

print(f" El valor del punto de la zona max de rechazo es {f_value_1}")

probability2 = 0.1  # The cumulative probability (e.g., for a 95% confidence level)
dfn2 = n1-1             # Numerator degrees of freedom (df1)
dfd2 = n2-1            # Denominator degrees of freedom (df2)

# Calculate the F value (critical value)
f_value_2 = stats.f.ppf(q=probability2, dfn=dfn2, dfd=dfd2)

print(f" El valor del punto de la zona min de rechazo es {f_value_2}")
```
