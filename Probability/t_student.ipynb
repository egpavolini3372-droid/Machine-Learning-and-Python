```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt


df = 16

x = np.linspace(-5,5,1000)

pdf_values = stats.t.pdf(x,df)

plt.plot(x,pdf_values)
plt.xlabel('x')
plt.ylabel('Probability Density Function')



```

```python
# Degrees of freedom
df = 288
# Significance level (alpha)
alpha = 0.975

# For a two-tailed test (e.g., 95% confidence):
# Probability q is 1 - alpha/2
q_two_tailed = 1 - alpha / 2
critical_t_two_tailed = stats.t.ppf(q_two_tailed, df)

print(f"Two-tailed critical t-value: {critical_t_two_tailed:.3f}")

# For a one-tailed test (e.g., right-tailed, alpha=0.05):
# Probability q is 1 - alpha
q_right_tailed = 1 - alpha
critical_t_right_tailed = stats.t.ppf(q_right_tailed, df)

print(f"Right-tailed critical t-value: {critical_t_right_tailed:.3f}")
```

```python
probability = 0.975  # The accumulated probability (e.g., for a 95% confidence interval, this is for one tail)
df = 6
# Get the t-value (critical value)
t_value = stats.t.ppf(probability, df)
print(t_value)
```

```python
# Define the degrees of freedom and the critical value
degrees_of_freedom = 16  # Example: degrees of freedom
critical_value = 2.11   # Example: t-statistic or critical value

# Calculate the cumulative probability (accumulated value)
cumulative_prob = stats.t.cdf(critical_value, df=degrees_of_freedom)

# Print the result
print(f"The cumulative probability for t-value of {critical_value} with {degrees_of_freedom} df is: {cumulative_prob}")
```
