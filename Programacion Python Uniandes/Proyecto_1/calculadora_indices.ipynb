```python

def calcular_IMC(peso,altura):
  IMC = peso/ (altura**2)
  return IMC

def calcular_porcentaje_grasa(peso,altura,edad,valor_genero):
  porc_grasa=(1.2*calcular_IMC(peso,altura))+(0.23*edad)-5.4-valor_genero
  return porc_grasa

def calcular_calorias_en_reposo(peso,altura,edad,valor_genero):
  TBM = (10*peso)+(6.25*altura*100)-(5*edad)+valor_genero
  return TBM

def calcular_calorias_en_actividad(peso, altura, edad, valor_genero,valor_actividad):
  TBM2 = calcular_calorias_en_reposo(peso,altura,edad,valor_genero)*valor_actividad
  return TBM2

def consumo_calorias_recomendado_para_adelgazar(peso,altura,edad,valor_genero):
  CCR1 = 0.8*calcular_calorias_en_reposo(peso,altura,edad,valor_genero)
  CCR2 = 0.85*calcular_calorias_en_reposo(peso,altura,edad,valor_genero)
  return(f"Para adelgazar es recomendado que consumas entre:{CCR1:.2f} y {CCR2:.2f} calorías al día. Siendo XXX el rango inferior y ZZZ el rango superior")

A =consumo_calorias_recomendado_para_adelgazar(88,1.78,35,10.8)
print(A)






```
