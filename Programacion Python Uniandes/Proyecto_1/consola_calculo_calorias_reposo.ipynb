```python
%%writefile consola_calculo_imc.py.py
def calcular_calorias_en_reposo(peso,altura,edad,valor_genero):
  peso = float(input("ingrese el peso de la persona (en kilogramos):"))
  altura = float(input("ingrese la altura de la persona (en metros):"))
  edad = float(input("ingrese la edad de la persona (en años):"))
  valor_genero = float(input("en caso de ser masculino debe ser 5 y en caso de ser femenino debe ser -161:"))

  TBM = (10*peso)+(6.25*altura*100)-(5*edad)+valor_genero
  return TBM
```
