```python
%%writefile consola_calculo_porcentaje_grasa.py
from consola_calculo_imc import calcular_IMC
def calcular_porcentaje_grasa(peso,altura,edad,valor_genero):
  peso = float(input("ingrese el peso de la persona (en kilogramos):"))
  altura = float(input("ingrese la altura de la persona (en metros):"))
  edad = float(input("ingrese la edad de la persona (en años):"))
  valor_genero = float(input("en caso de ser masculino debe ser 0 y en caso de ser femenino debe ser 10.8:"))
  porc_grasa=(1.2*calcular_IMC(peso,altura))+(0.23*edad)-5.4-valor_genero
  return porc_grasa
```
