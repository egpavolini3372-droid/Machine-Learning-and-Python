```python

capital = float(input("ingrese el capital a ahorrar: "))
interes = float(input("ingrese la tasa de interes ofrecida(%): "))
years = int(input("ingrese el tiempo de ahorro del capital: "))

ahorro = capital *((1+(interes/100))**years)
print(f"el capital de ahorro es {ahorro} a una tasa de {interes} % con un tiempo de {years} years")
```
