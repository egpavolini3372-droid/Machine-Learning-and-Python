```python
def calcular_cambio(cash_back):
  A = 500
  B = 200
  C = 100
  D = 50

  NA = 0
  NB = 0
  NC = 0
  ND = 0

  if cash_back >= A:
      NA = cash_back //500
      cash_back = cash_back % 500
  if cash_back >= B:
      NB = cash_back //200
      cash_back = cash_back % 200
  if cash_back >= C:
      NC = cash_back //100
      cash_back = cash_back % 100
  if cash_back >= D:
      ND = cash_back //50
  return(f"{NA },{NB },{ NC},{ND}")

calcular_cambio(1750)
```
