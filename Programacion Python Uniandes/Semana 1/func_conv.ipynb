```python
import math

def convertidor_fahr_cent(grad_fahr:float)->float:
  grad_cent = 5/9*(grad_fahr-32)

  return grad_cent

print(convertidor_fahr_cent(100))

def convertidor_cent_fahr(grad_cent):
  grad_fahr = 9/5*grad_cent+32
  return grad_fahr

print(convertidor_cent_fahr(37.7777))

def convertidor_rad_grados(rad):
  grados = rad*360/(2*math.pi)
  return grados

print(convertidor_rad_grados(math.pi))

def convertidor_grados_rad(grados):
  rad = grados*2*math.pi/(360)
  return rad

print(convertidor_grados_rad(180))

def invertir_numero(n: int) -> int:
    return int(str(n)[::-1])

print(invertir_numero(1234))

unidades = 3719 % 10
print(f"unidades {unidades}")
numero = 3719 // 10
print(f"numero1 {numero}")
decenas= numero % 10
print(decenas)
numero = numero // 10
print(numero)
centenas = numero % 10
print(centenas)
numero = numero // 10
print(numero)
millares = numero % 10
print(millares)

print(f"{unidades} {decenas} {centenas} {millares}")



```
