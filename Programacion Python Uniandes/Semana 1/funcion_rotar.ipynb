```python
def rotar():
  x1 = int(input(f"ingrese primer valor: "))
  x2 = int(input(f"ingrese segundo valor: "))
  x3 = int(input(f"ingrese tercer valor: "))

  x_1= x3
  x_2 = x1
  x_3 = x2
  print(f"el primer valor ahora es {x_1}", f"el segundo es {x_2}",  f"el tercero es {x_3}")

  return x_1,x_2,x_3

rotar()
```
