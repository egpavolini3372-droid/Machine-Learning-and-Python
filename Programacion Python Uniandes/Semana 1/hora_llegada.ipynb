```python
def calcular_horario_llegada(hora_salida,minutos_salida,segundos_salida,duracion_horas, duracion_minutos, duracion_segundos ):

  # --- Input Validation (independent checks) ---
  if hora_salida < 0 or hora_salida > 23:
    print("Hora de salida inválida")

  if minutos_salida < 0 or minutos_salida > 59:
    print("Minutos de salida inválidos")

  if segundos_salida < 0 or segundos_salida > 59:
    print("Segundos de salida inválidos")

  # Durations should typically be non-negative.
  # They can exceed 23 hours or 59 minutes/seconds as they represent a duration, not a point in time.
  if duracion_horas < 0:
    print("Duración en horas inválida: debe ser no negativa")

  if duracion_minutos < 0:
    print("Duración en minutos inválida: debe ser no negativa")

  if duracion_segundos < 0:
    print("Duración en segundos inválida: debe ser no negativa")

  # --- Calculation of arrival time components ---
  # These calculations were incorrectly indented before, leading to the UnboundLocalError.
  # They should always execute after initial checks.
  segundos_llegada = segundos_salida + duracion_segundos
  minutos_llegada = minutos_salida + duracion_minutos
  hora_llegada = hora_salida + duracion_horas

  # --- Normalize time components ---
  if segundos_llegada >= 60:
    minutos_adicionales = segundos_llegada // 60
    segundos_llegada = segundos_llegada % 60
    minutos_llegada += minutos_adicionales

  if minutos_llegada >= 60:
    hora_adicionales = minutos_llegada // 60
    minutos_llegada = minutos_llegada % 60
    hora_llegada += hora_adicionales

  if hora_llegada >= 24:
    hora_llegada = hora_llegada % 24

  # --- Output ---

  return f"{hora_llegada}:{minutos_llegada}:{segundos_llegada}"

calcular_horario_llegada(23,30,25,11,15,20)
```
