```python
def calcular_costo_boletas(cantidad_boletas, tipo_sala, hora_pico, pago_tarjeta_cinema, reserva):

    # Tarifas base
    if tipo_sala == "Dinamix":
        precio_base = 18800
    elif tipo_sala == "3D":
        precio_base = 15500
    else:  # 2D
        precio_base = 11300

    total = precio_base * cantidad_boletas

    # 🔹 Promoción 1: martes o miércoles, no hora pico, pago con tarjeta cinema
    if (not hora_pico) and pago_tarjeta_cinema:
        descuento = total * 0.10
        if cantidad_boletas > 3:
            descuento += 500 * cantidad_boletas
        total -= descuento

    # 🔹 Promoción 2: miércoles, no hora pico, sala 5D (nota: nunca aplica porque no está en tipos)
    if (not hora_pico) and tipo_sala == "5D":
        total -= 500 * cantidad_boletas

    # 🔹 Recargo por reserva
    if reserva:
        total += 2000 * cantidad_boletas

    # 🔹 Recargo hora pico
    if hora_pico:
        if tipo_sala == "Dinamix":
            total += 3000 * cantidad_boletas
        else:
            total += 2500 * cantidad_boletas

    return round(total)

print(calcular_costo_boletas(1, "2D", False, False, False))
```
