```python
def mas_a(c1:str,c2:str,c3:str,c4:str)->str:
  letra = "a"
  cadenas_mas =c1 # se inicializa la variable con la primera cadena
  cantidad_mas = c1.lower().count(letra) # se cuenta el numero de letras "a" de la cadena

  if c2.lower().count(letra) > cantidad_mas:
    cadenas_mas = c2
    cantidad_mas = c2.lower().count(letra)
  if c3.lower().count(letra) > cantidad_mas:
    cadenas_mas = c3
    cantidad_mas = c3.lower().count(letra)
  if c4.lower().count(letra) > cantidad_mas:
    cadenas_mas = c4
    cantidad_mas = c4.lower().count(letra)

  return cadenas_mas, cantidad_mas



print(mas_a("carro", "casa", "acampar", "amarillita"))

```
