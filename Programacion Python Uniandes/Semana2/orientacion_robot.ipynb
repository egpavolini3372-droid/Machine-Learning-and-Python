```python
def movimiento_robot(orientacion_actual:str, giro_1:str, giro_2:str, giro_3:str):
  list = ["W","N","E","S","W","N","E","S","W","N","E","S","W","N","E","S"]

  if orientacion_actual == "N":
    pos = list[1]
  elif orientacion_actual == "S":
    pos = list[3]
  elif orientacion_actual == "E":
    pos = list[2]
  elif orientacion_actual == "W":
    pos = list[0]

  if giro_1 == "L":
    pos = list[list.index(pos)-1]
  elif giro_1 == "R":
    pos = list[list.index(pos)+1]
  elif giro_1 == "H":
    pos = list[list.index(pos)+2]
  elif giro_1 == """ "." """:
    pos = list[list.index(pos)+0]


  if giro_2 == "L":
    pos = list[list.index(pos)-1]
  elif giro_2 == "R":
    pos = list[list.index(pos)+1]
  elif giro_2 == "H":
    pos = list[list.index(pos)+2]
  elif giro_2 == """ "." """:
    pos = list[list.index(pos)+0]


  if giro_3 == "L":
    pos = list[list.index(pos)-1]
  elif giro_3 == "R":
    pos = list[list.index(pos)+1]
  elif giro_3 == "H":
    pos = list[list.index(pos)+2]
  elif giro_3 == """ "." """:
    pos = list[list.index(pos)+0]

  return pos

print( movimiento_robot("S", "R", "R", "H"))
```
