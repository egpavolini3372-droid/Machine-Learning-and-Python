```python
def picas_y_fijas(numero_secreto:int,intento:int):

  dicc = {'PICAS':0,'FIJAS':0}

  digito_i_3 = intento % 10
  intento_3 = intento // 10
  digito_i_2 = intento_3 % 10
  intento_2 = intento_3 // 10
  digito_i_1 = intento_2 % 10
  intento_1 = intento_2 // 10
  digito_i_0 = intento_1 % 10
  intento_0 = intento_1 // 10

  digito_ns_3 = numero_secreto % 10
  numero_secreto_3 = numero_secreto // 10
  digito_ns_2 = numero_secreto_3 % 10
  numero_secreto_2 = numero_secreto_3 // 10
  digito_ns_1 = numero_secreto_2 % 10
  numero_secreto_1 = numero_secreto_2 // 10
  digito_ns_0 = numero_secreto_1 % 10
  numero_secreto_0 = numero_secreto_1 // 10


  if digito_i_3 == digito_ns_3:
    dicc['fijas'] += 1
  if digito_i_3 == digito_ns_2 or digito_i_3 == digito_ns_1 or digito_i_3 == digito_ns_0:
    dicc['picas'] += 1
  if digito_i_2 == digito_ns_2:
    dicc['fijas'] += 1
  if digito_i_2 == digito_ns_3 or digito_i_2 == digito_ns_1 or digito_i_2 == digito_ns_0:
    dicc['picas'] += 1
  if digito_i_1 == digito_ns_1:
    dicc['fijas'] += 1
  if digito_i_1 == digito_ns_3 or digito_i_1 == digito_ns_2 or digito_i_1 == digito_ns_0:
    dicc['picas'] += 1
  if digito_i_0 == digito_ns_0:
    dicc['fijas'] += 1
  if digito_i_0 == digito_ns_3 or digito_i_0 == digito_ns_2 or digito_i_0 == digito_ns_1:
    dicc['picas'] += 1
  return dicc

print(picas_y_fijas(1234,1325))
```
