```python
def mismos_digitos(a, b):

    sa = str(a)
    sb = str(b)

    # identificar el número con menos y más dígitos
    if len(sa) <= len(sb):
        menor = sa
        mayor = sb
    else:
        menor = sb
        mayor = sa

    i = 0
    while i < len(menor):
        encontrado = False
        j = 0

        while j < len(mayor):
            if menor[i] == mayor[j]:
                encontrado = True
            j += 1

        if encontrado == False:
            return False

        i += 1

    return True
print(mismos_digitos(89,998))
```
