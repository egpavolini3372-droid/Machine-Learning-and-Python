```python
def sucesion_fibonacci(cantidad_numeros: int) -> list:
  if cantidad_numeros <= 0:
    return []
  elif cantidad_numeros == 1:
    return [0]

  fibonacci = [0, 1]
  # Start the loop from index 2, as 0 and 1 are already in the list
  i = 2
  while i < cantidad_numeros:
    next_num = fibonacci[i-1] + fibonacci[i-2]
    fibonacci.append(next_num)
    i += 1

  return fibonacci

print(sucesion_fibonacci(12))
```

```python
def sucesion_fibonacci_str(cantidad_numeros: int) -> str:
  fibonacci_list = sucesion_fibonacci(cantidad_numeros)
  # Convert the list of numbers to a string, joined by commas and spaces
  fibonacci_string = ", ".join(map(str, fibonacci_list))
  return fibonacci_string

print(sucesion_fibonacci_str(12))
```

```python
def sucesion_fibonacci_str_from_scratch(cantidad_numeros: int) -> str:
  if cantidad_numeros <= 0:
    return ""
  elif cantidad_numeros == 1:
    return "0"

  fibonacci = [0, 1]
  i = 2
  while i < cantidad_numeros:
    next_num = fibonacci[i-1] + fibonacci[i-2]
    fibonacci.append(next_num)
    i += 1

  # Convert the list of numbers to a string, joined by commas and spaces
  fibonacci_string = ",".join(map(str, fibonacci))
  return fibonacci_string

print(sucesion_fibonacci_str_from_scratch(3))
```
