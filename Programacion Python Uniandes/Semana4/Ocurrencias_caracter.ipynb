```python
def ocurrencias_caracter(cadena : str, caracter:str)-> int:

  ocurrencias = 0

  for cada_caracter in cadena:
    if (cada_caracter == caracter):
       ocurrencias +=1

  return ocurrencias

print(ocurrencias_caracter("La casa Blanca", "a"))
```

```python
def calculate_sum(a:int)->int:
  resultado = 0
  for i in range(0, a+1):
    resultado +=i

  return resultado

print(calculate_sum(4))
```

```python
def contar_caracteres_repetidos(cadena: str) -> int:
    conteo = {}

    # Contar frecuencia de cada carácter en un diccionario {caracter: # repeticiones}
    for c in cadena:
        if c in conteo:
            conteo[c] += 1

        else:
            conteo[c] = 1

    # Contar caracteres con más de una aparición
    repetidos = 0
    for valor in conteo.values():
        if valor > 1:
            repetidos += 1

    return repetidos

print(contar_caracteres_repetidos("Hola mundo"))
```

```python
def ordenar_cadena(cadena: str) -> str:

  ordenado = "".join(sorted(cadena,key=str.lower))

  return ordenado
print(ordenar_cadena("Hola mundo"))
```

```python
def letra_mas_comun(cadena: str) -> int:
    conteo = {}

    # Contar frecuencia de cada carácter en un diccionario {caracter: # repeticiones}
    for c in cadena:
        if c in conteo:
            conteo[c] += 1

        else:
            conteo[c] = 1

    # Contar caracteres con más de una aparición
    repetidos = 0
    for valor in conteo.values():
        if valor > 1:
            repetidos += 1

    max_frecuencia = -1
    resultado = ""

    for letra in conteo:
        if conteo[letra] > max_frecuencia:
            max_frecuencia = conteo[letra]
            resultado = letra
        elif conteo[letra] == max_frecuencia:
            # desempate: letra alfabéticamente posterior
            if letra > resultado:
                resultado = letra

    return resultado

print(letra_mas_comun("Hola mundo soy yo"))
```
