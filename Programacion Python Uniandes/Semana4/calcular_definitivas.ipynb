```python
def calcular_definitivas(estudiantes:list)->list:

  notas_aprox = []

  for alumno in estudiantes:
    if alumno["nota"] >= 4.5:
      new_nota = 5.0

    elif 3.5 <= alumno["nota"] < 4.5:
      new_nota = 4.0
    elif 2.5 <= alumno["nota"] < 3.5:
      new_nota = 3.0
    else:
      new_nota = 1.5

    notas_aprox.append({"nombre": alumno["nombre"], "nota": new_nota})
  return notas_aprox

estudiantes = [{'nombre': 'Juan', 'nota': 2.7}, {'nombre': 'Juanito', 'nota': 4.55}, {'nombre': 'Pedro', 'nota': 3.4}, {'nombre': 'Mateo', 'nota': 1.9}, {'nombre': 'Andrea', 'nota': 4.24}]
alumnos = [
    {"nombre": "Ana", "nota": 4.8},
    {"nombre": "Juan", "nota": 3.7},
    {"nombre": "Pedro", "nota": 2.1}
]

print(calcular_definitivas(estudiantes))
```
