```python
def promedio_lista(numeros:list)->float:
  """ Funcion que suma los valores POSITIVOS de una lista y los divide por el numero de valores POSITIVOS"""
  sumatoria = 0
  total = 0

  for num in numeros:
    if num > 0:
      sumatoria += num
      total += 1

  return sumatoria / total

a = [1,2,3,4,-1,-2,-3,-4,4,4]
print(promedio_lista(a))
```

```python
def menor_posicion_lista(numeros:list)->int:
  """ Funcion que devuelve la posicion del menor valor de una lista"""
  menor = numeros[0]
  for num in numeros [1:]:
    if num < menor:
      menor = num
      posicion = numeros.index(menor)

  return posicion

print(menor_posicion_lista(a))
```

```python
def buscar_elemento(numbers:list, number:int)->int:
   """ función que recibe una lista y un número entero a buscar, y retorna un entero que indica el índice en que se encuentra este elemento"""

   if number not in numbers:
    return -1
   else:
    buscado = numbers.index(number)

   return buscado

print(buscar_elemento(a, 4))
```

```python
def encontrar_mayor(entrada:list)->int:

  if entrada == []:
    mayor = -1
  else:
    mayor = entrada[0]
    for i in entrada[1:]:
          if i > mayor:
                  mayor = i

  return mayor
print(encontrar_mayor(a))
```

```python
def insertar_ordenado(lista_ordenada:list, cadena:str)-> list:
  i = 0

  while i < len(lista_ordenada) and cadena > lista_ordenada[i]:

    i += 1


  lista_ordenada.insert(i, cadena)

  return lista_ordenada

lista_paises = ["Algeria","Argentina", "Bolivia","Colombia", "Perú", "Venezuela"]
print(insertar_ordenado(lista_paises, "Brasil"))
```

```python
def intercalar_palabras(cad1:str, cad2:str)->str:
  resultado=""

  palabras_1 = cad1.split()
  palabras_2 = cad2.split()

  for i in range(min(len(palabras_1), len(palabras_2))):
    resultado += palabras_1[i] + " " + palabras_2[i] + " "
  return resultado

print(intercalar_palabras("Hola hermoso gigante que el qué", "Mundo y Tierra respira oxígeno"))
```
