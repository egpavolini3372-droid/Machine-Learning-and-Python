```python
def producto_mas_costoso(carrito_compras:dict)->str:

   if not carrito_compras:
        return "No hay productos en el carrito"

   # Sort items by price in descending order, then by name alphabetically in ascending order.
   # The lambda function creates a tuple for each item: (negative_price, lowercased_name).
   # Sorting by negative price makes the highest price come first.
   # Sorting by lowercased name (second element) alphabetically breaks ties.
   sorted_items = sorted(carrito_compras.items(), key=lambda item: (-item[1], item[0].lower()))

   # The first element in the sorted list will be the product with the highest price,
   # and if there's a tie, the one that comes first alphabetically.
   return sorted_items[0][0]

carrito_compras = {'bananos': 2400, 'chocolatinas': 4000, 'detergente': 4000}

print(producto_mas_costoso(carrito_compras))
print(carrito_compras.items())
```
