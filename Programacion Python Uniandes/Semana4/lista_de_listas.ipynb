```python
def hay_negativo(matriz:list)->bool:
  for row in matriz:
    for element in row:
      if element < 0:
        return True
  return False

matriz = [[1,2,4],[4,2,8],[7,1,2]]
print(hay_negativo(matriz))
```

```python
def pintar_x(matriz:list, operacion:str)->list:
  n = len(matriz)

  # Process main diagonal
  for i in range(n):
    a = matriz[i][i]
    if operacion == "+":
      matriz[i][i] = a + a
    elif operacion == "*":
      matriz[i][i] = a * a
    elif operacion == "-":
      matriz[i][i] = a - a
    elif operacion == "/":
      if a != 0:
        matriz[i][i] = a / a
      else:
        matriz[i][i] = 0 # Avoid ZeroDivisionError, assume 0/0 is 0

  # Process anti-diagonal, skipping the central cell if n is odd
  for i in range(n):
    # If n is odd and i is the middle index, this is the central cell
    # which has already been processed by the main diagonal loop. So, skip it.
    if n % 2 == 1 and i == n // 2:
      continue

    b = matriz[i][n - 1 - i]
    if operacion == "+":
      matriz[i][n - 1 - i] = b + b
    elif operacion == "*":
      matriz[i][n - 1 - i] = b * b
    elif operacion == "-":
      matriz[i][n - 1 - i] = b - b
    elif operacion == "/":
      if b != 0:
        matriz[i][n - 1 - i] = b / b
      else:
        matriz[i][n - 1 - i] = 0 # Avoid ZeroDivisionError, assume 0/0 is 0

  return matriz


matriz7 = [[2,2,2],[2,2,2],[2,2,2]]
pintar_x(matriz7,"*")
```
