```python
def find_words_in_text(words: list[str], text: str) -> dict[str, tuple[int, int, int]]:
    """
    Finds specific words in a text and collects statistics about each.

    Args:
        words: List of words to search for (case-insensitive).
        text: The text to search in.

    Returns:
        A dictionary where each key is a word (lowercase) and the value
        is a tuple of (count, first_position, last_position).
        Words not found in the text are not included in the result.
    """
    result = {}
    text_words = text.lower().split()

    for word in words:
        word_lower = word.lower()
        positions = [i for i, w in enumerate(text_words) if w == word_lower]

        if positions:
            count = len(positions)
            first_pos = positions[0]
            last_pos = positions[-1]
            result[word_lower] = (count, first_pos, last_pos)

    return result

text = "The cat sat on the mat and the cat looked at the mat"
words = ["cat", "mat", "dog", "the"]
print(find_words_in_text(words, text))

```

```python
def analizar_texto(texto:str, caracteres_permitidos:list) -> dict:

    """
    Finds specific words in a text and collects statistics about each.

    Args:
        words: List of words to search for (case-insensitive).
        text: The text to search in.

    Returns:
        A dictionary where each key is a word (lowercase) and the value
        is a tuple of (count, first_position, last_position).
        Words not found in the text are not included in the result.
    """
    result = {}
    text_lower = texto.lower()

    for word in caracteres_permitidos:
        word_lower = word.lower()
        word_len = len(word_lower)
        positions = []
        i = 0

        while i <= len(text_lower) - word_len:
            if text_lower[i:i + word_len] == word_lower:
                # Check word boundaries manually
                before = (i == 0) or (not text_lower[i - 1].isalpha())
                after = (i + word_len == len(text_lower)) or (not text_lower[i + word_len].isalpha())

                if before and after:
                    positions.append(i)

            i += 1

        if positions:
            result[word_lower] = (len(positions), positions[0], positions[-1])

    return result

text = "The cat sat on the mat and the cat looked at the mat"
words = ["cat", "mat", "dog", "the"]
print(analizar_texto(text,words))
```
