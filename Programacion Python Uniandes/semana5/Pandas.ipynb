```python
def subir_notas(notas:pd.Series)->pd.Series:
  if notas.mean()> 2.5:
    notas = notas + round(notas.std(),2)

  for i in range(notas.size):
    nota_actual = notas.iloc[i]
    if nota_actual > 5:
      notas.iloc[i] = 5

  return notas

notas = pd.Series(np.random.normal(3.2, 1.2, 40))
print(subir_notas(notas))

```

```python
import pandas as pd
def mejores_estudiantes(estudiantes: pd.DataFrame):

  promedio = estudiantes[['matematicas','ingles', 'ciencias', 'literatura', 'arte' ]].mean(axis=1)
  estudiantes['promedio'] = promedio
  ordenado = estudiantes.sort_values("promedio", ascending=False)

  n = len(ordenado)
  top_25 = int(n*0.25)

  best_25 = ordenado.head(top_25)

  est_prom_ord= ordenado[['nombre','promedio']].round(2)
  best25 = ordenado.head(top_25)

  best_25["promedio"] = best_25["promedio"].round(2)

  resultado = best_25[["nombre", "promedio"]]

  return resultado

```

```python
def funcion(t, inc) -> int:
  v1, v2, v3 = t
  rt = v1 + inc, v2+inc, v3*inc
  return rt


print(funcion((4,7,9), 3))
```

```python
def funcion(t:tuple, f:int, e:int)-> int:
  r= 0
  i = 0
  while i < len(t):
    j = 0
    while j < len(t[i]):
      r += t[i][j] *f + e
      j +=1
    i +=1
  return r

var = ((1,2),(3,5),(-3),(42,28))
print(funcion(var,3,5))
```

```python
def funcion(t:tuple, f:int)-> int:

  res=[]
  for v1, v2 in t:
    res.append(v1*f + v2)
  return res

var = ((1,2),(3,5),(-3,-2),(42,28))
print(funcion(var,3))
```
