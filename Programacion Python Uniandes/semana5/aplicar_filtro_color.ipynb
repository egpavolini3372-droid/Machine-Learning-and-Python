```python
def aplicar_filtro_color(imagen:list, conservar:tuple)->list:
    nueva_imagen = []
    alto = len(imagen)
    ancho = len(imagen[0])

    for i in range(alto):
      fila_nueva = []
      for j in range(ancho):
        r,g,b = imagen[i][j]
        temp = [r,g,b]
        for k in range(3):
          if not conservar(k):
            temp[k]= 0.0
        nuevo_pixel ={temp[0],temp[1],temp[2]}
        fila_nueva.append(nuevo_pixel)
      nueva_imagen.append(fila_nueva)
    return nueva_imagen
```

```python
def binarizar_imagen(imagen:list, umbral:float)->list:
  nueva_imagen = []
  alto = len(imagen)
  ancho = len(imagen[0])

  for fila in imagen:
    fila_nueva = []
    for pixel in fila:
      promedio = sum(pixel)/3

      if promedio >= umbral:
        fila_nueva.append((1,1,1))
      else:
        fila_nueva.append((0,0,0))
    nueva_imagen.append(fila_nueva)

    return nueva_imagen



```

```python
def binarizar_imagen(imagen, umbral):
    imagen_binarizada = []

    for fila in imagen:
        nueva_fila = []
        for pixel in fila:
            promedio = sum(pixel) / 3

            if promedio >= umbral:
                nueva_fila.append((1, 1, 1))
            else:
                nueva_fila.append((0, 0, 0))

        imagen_binarizada.append(nueva_fila)

    return imagen_binarizada
```

```python
def reflejar_imagen(imagen:list)->list:
  imagen_reflejada = []


  for fila in imagen:
      fila_nueva = fila[::-1]
      imagen_reflejada.append(fila_nueva)

  return imagen_reflejada

```
