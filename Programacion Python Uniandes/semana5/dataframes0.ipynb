```python
import pandas as pd
import numpy as np

def depurar_partidos(partidos: pd.DataFrame) -> pd.DataFrame:
    df = partidos.copy()

    # 1. Reemplazar valores nulos en goles por 0
    df['goles_local'] = df['goles_local'].fillna(0)
    df['goles_visitante'] = df['goles_visitante'].fillna(0)

    # 2. Corregir la columna 'resultado' según los goles reales
    def calcular_resultado(row):
        if row['goles_local'] > row['goles_visitante']:
            return row['local']
        elif row['goles_visitante'] > row['goles_local']:
            return row['visitante']
        else:
            return 'empate'

    df['resultado'] = df.apply(calcular_resultado, axis=1)

    # 3. Eliminar partidos donde un equipo jugó contra sí mismo
    df = df[df['local'] != df['visitante']]

    # Mantener el orden original con reset_index
    df = df.reset_index(drop=True)

    return df
```

```python
import pandas as pd

def calcular_habitantes_por_puesto(poblacion, universidades):

    # 1. Agrupar estudiantes por país
    estudiantes_por_pais = universidades.groupby('country', as_index=False)['num_students'].sum()

    # 2. Merge con población
    df = poblacion.merge(estudiantes_por_pais, left_on='Pais', right_on='country', how='inner')

    # 3. Calcular habitantes por puesto (sin redondear aún)
    df['habitantes_por_puesto'] = df['Poblacion'] / df['num_students']

    # 4. Ordenar antes de redondear
    df = df.sort_values(by='habitantes_por_puesto', ascending=True)

    # 5. Redondear a 1 decimal
    df['habitantes_por_puesto'] = df['habitantes_por_puesto'].round(1)

    # 6. Seleccionar columnas finales
    resultado = df[['Pais', 'habitantes_por_puesto']].reset_index(drop=True)

    return resultado
```
