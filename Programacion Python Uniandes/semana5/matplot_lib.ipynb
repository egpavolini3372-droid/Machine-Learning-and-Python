```python
import matplotlib.pyplot as plt
import numpy as np


meses = np.array(["Ene", "Feb","Mar", "Abr", "May", "Jun", "Jul", "Ago", "Sep", "Oct", "Nov", "Dic"])
temp = np.array([14.3, 14.3, 15, 14.7, 14.7, 14.5, 14.3, 14.2, 14.3, 14.3, 14.5, 14.4])

plt.figure(figsize = (7,4))
plt.plot(meses, temp, label ="temperatura")
plt.xlabel("meses")
plt.ylabel("Temperatura [°C]")
plt.title("Temperatura promedio mensual en Bogota (1970-2000)")
plt.legend()
plt.show()

#
```

```python
fig = plt.figure(figsize = (7,4))
ax = fig.add_axes([0.1,0.1,0.8,0.8]) # primero numero: posicion limite izquierdo con respecto a la figura (10% a la izquierda), lo mimso abajo, ancho, lo mismo alto
ax.plot(meses, temp , label ="temperatura")
ax.set_xlabel("Meses")
ax.set_ylabel("Temperatura [°C]")
ax.set_title("Temperatura Promedio en Bogota, por mes (1970-2000)")
ax.legend()
fig.savefig("imagen.png")
```

```python
import random
notas = []
ids_estudiantes = []

for i in range(50):
  nota_generada = random.normalvariate(3.25,0.98)
  nota_generada = round(nota_generada,2)
  agregado = False
  while not agregado:
    id_estudiante = random.randint(20202001, 20202999)
    id_generado = str(id_estudiante)
    if id_generado not in ids_estudiantes:
      notas.append(nota_generada)
      ids_estudiantes.append(id_generado)
      agregado = True


plt.plot(ids_estudiantes, notas)
plt.title("Notas del curso")
plt.ylabel("Nota")
plt.xlabel("ID del estudiante")
plt.xticks(rotation=90, fontsize=7)
plt.show()

plt.hist(notas, bins=12, color = ["green"], edgecolor="black", linewidth=1)
plt.title("Cantidad de estudiantes por rango de nota")
plt.ylabel("Cantidad de estudiantes")
plt.xlabel("Rango de nota")
plt.show()

plt.scatter(ids_estudiantes, notas)
plt.title("Nota vs ID")
plt.ylabel("Nota")
plt.xlabel("ID)")
plt.xticks(rotation=90, fontsize=7)
plt.show()
```
