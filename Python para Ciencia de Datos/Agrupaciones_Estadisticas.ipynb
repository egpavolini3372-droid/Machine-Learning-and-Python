```python
import pandas as pd

path = '/content/Online_Retail.csv'
df = pd.read_csv(path, encoding='latin1')

df["TotalPrice"] = df["Quantity"]* df["UnitPrice"]

print(df.head())
```

```python
country_count = df["Country"].value_counts()
print(country_count)
```

```python
# La funcion .groupby() permite agrupar por la suma total de la cantidad
# Si la celda dice Country = "Germany" y la "Quantity" es 7, va a sumar 7 para Germany
country_group = df.groupby("Country")["Quantity"].sum()
print(country_group)
```

```python
# .agg() adiciona información  basado en "Country" y "UnitPrice"ù
country_stats = df.groupby("Country")["UnitPrice"].agg(["mean","sum"])
print(country_stats)
```

```python
# Agrupar por Pais y sub agrupar por Stockcode, sumando los valores
country_stock_group = df.groupby(["Country","StockCode"])["Quantity"].sum()
print(country_stock_group)
```

```python
# Agrupar por Pais y sub agrupar por Stockcode, sumando los valores
country_stock_group = df.groupby(["StockCode","Country"])["Quantity"].sum()
print(country_stock_group)
```

```python
# Se define una funcion para calcular el revenue: Cantidad * precio unitario.
# El revenue por country agrupa los paisis y aplica la funcion creada anteriormente

def total_revenue(group):
  return (group["Quantity"]* group['UnitPrice']).sum()

revenue_per_country = df.groupby("Country").apply(total_revenue)
print(revenue_per_country)
```

```python
country_rank = revenue_per_country.sort_values(ascending=False)


print(country_rank)
top3_best = country_rank.iloc[:3]
print("Los 3 países con mayores ventas: primeros en el ranking:\n", top3_best)


top3_worst = country_rank.iloc[-3:]
print("Los 3 países con menores ventas: últimos en el ranking:\n", top3_worst)
```
