```python
import numpy as np
```

```python
array = np.array([[1,2,3],[4,5,6]])
print(array.ndim)
print(array.shape)
print(array.dtype)
```

```python
z = np.array(3, dtype=np.uint8)
print(z)
```

```python
double_array = np.array([1,2,3], dtype= "d")
print(double_array)
```

```python
z = z.astype(np.float64)
print(z)
```

```python
sum = np.sum(array)
print(f"the sum is {sum}")

mean = np.mean(array)
print(f"the mean is {mean}")

desv_stan =np.std(array)
print(f"the standar deviation is {desv_stan}")
```

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
print(A+B)
print(A-B)
print(A*B)
print(A/B)
```

```python
producto = np.dot(A, B)
print("Producto de matrices:\n", producto)
```

```python
determinante = np.linalg.det(A)
print("Determinante de A:", determinante)
```

```python
trans_A = A.T
print("Transpuesta de A:", trans_A)
```

```python
inversa = np.linalg.inv(A)
print("Inversa de A:\n", inversa)
```

```python
valores_propios, vectores_propios = np.linalg.eig(A)
print("Valores propios de A:\n", valores_propios)
print("Vectores propios de A:\n", vectores_propios)
```

```python
B = np.array([1, 2])
X = np.linalg.solve(A, B)
print("Solución del sistema AX = B:\n", X)
```
