```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(165, 15, 1000)
print(data)

plt.figure(figsize=(8,6))
```

```python
# Alpha = transparencia de las barras, bins = numero de intervalos
plt.hist(data, bins=20, color='salmon', edgecolor="black", alpha=0.7)
plt.title("Distribucion Normal de estaturas")
plt.xlabel("Altura en cm")
plt.ylabel("Frecuencia")
```

```python
# con seed permanecen constantes los numeros aleatorios que se generan la primera vez
np.random.seed(0)
ages = [np.random.normal(35, 25, 1000),np.random.normal(41, 35, 1000),np.random.normal(50, 50, 1000)]
#
print(ages)
```

```python
#
plt.boxplot(ages, patch_artist=True, notch= False, vert=True, labels=["Group1", "Group2","Group3"])
plt.title("Distribucion de edades por grupo")
plt.xlabel("Grupo")
plt.ylabel("Edad en años")
```
