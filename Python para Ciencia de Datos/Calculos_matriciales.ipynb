```python
import numpy as np

A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
C = A + B
print(A)
print(B)
print(C)
```

```python
# Multiplicacion

product = np.dot(A,B)
print(product)
```

```python
# Determinantes e Inversa
det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)
print(det_A)
print(inv_A)
```

```python
# Resolver sistema de ecuaciones Ax = b

b = np.array([9,11])
x = np.linalg.solve(A,b)
print(x)

```
