```python
# .isna() funcion para identificar celdas faltantes o vacias
missing_data = retail_data.isna()
print(missing_data.head())
```

```python
# isna().sum() funciones para sumar las celdas faltantes por columna
missing_data_count = retail_data.isna().sum()
print(f"conteo de datos por columna:\n {missing_data_count}")
```

```python
# Eliminar filas de datos faltantes, dropna() funcion que elimina las filas de los datos faltantes
no_missing_data = retail_data.dropna()
print(no_missing_data)
```

```python
# Eliminar las columnas de datos faltantes dropna(axis =1)

no_missing_columns = retail_data.dropna(axis=1)
print(no_missing_columns)
```

```python
#Remplaza los valores restantes con ceros fillna(0)
# isna en este caso es cero dado que se llenaron las celdas vacias con ceros
retail_data_fill_zeros = retail_data.fillna(0)
retail_data_fill_zeros_count = retail_data_fill_zeros.isna().sum()
print(retail_data_fill_zeros_count)
#
```

```python
# Variable["Llave"].mean() devuelve la media de una columna

mean_unit_values = retail_data['UnitPrice'].mean()
print(mean_unit_values)
# rellena los valores faltantes con la media

retail_data_filled_mean =  retail_data['UnitPrice'].fillna(mean_unit_values)
print(retail_data_filled_mean)
```

```python
# Eliminar filas de datos faltantes, dropna() funcion que elimina las filas de los datos faltantes
no_missing_data = retail_data.dropna(subset=["CustomerID", "InvoiceDate"], inplace =True)
print(no_missing_data)
```
