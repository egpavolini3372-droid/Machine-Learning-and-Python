```python
import pandas as pd

path = '/content/Online_Retail.csv'
df = pd.read_csv(path, encoding='latin1')

df["TotalPrice"] = df["Quantity"]* df["UnitPrice"]

print(df.head())
```

```python
df["HighValue"]= df['TotalPrice'] > 20
print(df["HighValue"].head(30))
```

```python
print(df.info())
```

```python
# Cambia el tipo de datos de la columna "InvoiceDate" de "object" a "datetime64"
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
print(df.info())
print(df.head())
```

```python
# crear una nueva columna con descuentos del precio unitario usando una funcion lamda
df['DiscountPrice'] = df["UnitPrice"].apply(lambda x: x*0.9)
print(df.head())
```

```python
def categorizar_price(price):
  if price > 50:
    return "High"
  elif price > 10:
    return "Medium"
  else:
    return "Low"

df["PriceCategory"] = df["UnitPrice"].apply(categorizar_price)
print(df["PriceCategory"])
```
