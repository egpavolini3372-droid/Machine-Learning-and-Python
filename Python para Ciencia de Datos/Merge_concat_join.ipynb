```python
import pandas as pd
# Crear DataFrames de ejemplo
df1 = pd.DataFrame({
    'key': ['A', 'B', 'C'],
    'value1': [1, 2, 3]
})

df2 = pd.DataFrame({
    'key': ['B', 'C', 'D'],
    'value2': [4, 5, 6]
})

print(df1)
print(df2)

# Realizar un merge interno (copia las columnas que tienen valores de Keys iguales)
# en este caso B= 2 y B = 4 AND C =3 y C =5
inner_merged = pd.merge(df1, df2, on='key', how='inner')
print("Unión Interna:\n", inner_merged)
```

```python
# Realiza un merge externo donde muestra los valores de todas las keys disponibles en ambos data Frames
# Independiente de si se repiten
# Llena con NaN donde no hay valor especificado en alguno de los data frames
outer_merged = pd.merge(df1,df2, on = "key", how = "outer")
print("Unión Externa:\n", outer_merged)
```

```python
# toma los valores del data frame de la izquiera "df1" solo se toma en cuenta los que coinciden con el otro data frame
left_merged = pd.merge(df1, df2, on = "key", how ="left")
print(left_merged)
```

```python
# toma los valores del data frame de la izquiera "df2" solo se toma en cuenta los que coinciden con el otro data frame
right_merged = pd.merge(df1, df2, on = "key", how ="right")
print(right_merged)
```

```python
# Ejercicios con concat()
df3 = pd.DataFrame({
  'A': ['A0', 'A1', 'A2'],
  'B': ['B0', 'B1', 'B2'],
})
df4 = pd.DataFrame({
  'A': ['A3', 'A4', 'A5'],
  'B': ['B3', 'B4', 'B5'],
})

print(df3)
print(df4)
```

```python
vertical_concat =pd.concat([df3, df4])
print(vertical_concat)
```

```python
horizon_concat =pd.concat([df3, df4],axis = 1)
print(horizon_concat)
```

```python
# Ejercicios con join()
df5 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2']
  },
  index = ['K0', 'K1', 'K2']
)
df6 = pd.DataFrame({
    'C': ['C0', 'C1', 'C2'],
    'D': ['DO', 'D1', 'D2']
  },
  index = ['K0', 'K2', 'K3']
)
print(df5)
print(df6)
```

```python
# Union de dos data frames, solo toma los indices que coinciden K0 y K2
joined = df5.join(df6, how = "inner")
print(joined)
```
