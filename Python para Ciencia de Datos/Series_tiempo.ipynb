```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.dates import DateFormatter

dates = pd.date_range(start = "2023-01-01", periods = 100)
values = np.random.rand(100).cumsum()

data = pd.DataFrame({"Date": dates, "Value": values})

#Crear Grafico de Lineas

fig, ax = plt.subplots(figsize=(12,6))

ax.plot(data["Date"], data["Value"], color = "green")

plt.xticks(rotation=45)
plt.title("Series de Tiempo ")
plt.xlabel("Fecha")
plt.ylabel("Valor")

plt.show()

#plt.plot(data["Date"], data["Value"])


```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.dates import DateFormatter

dates = pd.date_range(start = "2023-01-01", periods = 12, freq="M")
sales = np.random.randint(1000,5000, size=12)

sales_data = pd.DataFrame({"Date": dates, "Sales": sales})

#Crear Grafico de Lineas

plt.plot(sales_data["Date"], sales_data["Sales"], marker="o",linestyle="-",label="Ventas mensuales")

plt.gca().xaxis.set_major_formatter(DateFormatter("%b %Y"))

# rotacion de los valores del eje x
plt.xticks(rotation=45)

plt.title("Análisis de ventas mensuales ")
plt.xlabel("Fecha")
plt.ylabel("Ventas")

# Leyenda dentro dle gráfico
plt.legend()

# Matplotlib automatically adjusts subplot parameters to give specified padding, ensuring that plot elements like axes labels and titles do not overlap and fit neatly within the figure area
plt.tight_layout()

plt.show()
```
