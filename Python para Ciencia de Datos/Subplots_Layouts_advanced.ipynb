```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.gridspec import GridSpec

x = np.linspace(0,10,100) # primer valor: inicio, segundo: final, tercero:numero de puntos
y = np.sin(x)
z = np.cos(x)

data = np.random.randn(100)
```

```python
gs = GridSpec(2,2, height_ratios=[2,1], width_ratios=[1,1])
fig = plt.figure(figsize=(10,8))

# Primer Subplot grande (ocupa toda la primera fila)

ax1 = fig.add_subplot(gs[0,:])
ax1.plot(x,y,color = "blue")
ax1.set_title("Seno de x")
ax1.set_xlabel("x")
ax1.set_ylabel("sin(x)")

# segundo subplot, en la esquina inferior-izquierda

ax2 = fig.add_subplot(gs[1,0])
ax2.hist(data, bins=20, color="green",edgecolor = "black")
ax1.set_title("Histograma")
ax1.set_xlabel("Valor")
ax1.set_ylabel("Frecuencia")

# Tercer subplot, ubicado en la esquina inferior derecha

ax3 = fig.add_subplot(gs[1,1])
ax3.scatter(x,z,color = "red")
ax1.set_title("Coseno de x")
ax1.set_xlabel("x")
ax1.set_ylabel("cos(x)")

plt.tight_layout()
plt.show()
```
