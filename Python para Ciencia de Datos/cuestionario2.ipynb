```python
from scipy.stats import norm
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from scipy import stats
from statistics import NormalDist

# Data arrays 1 and 2
data1 =  np.array([409,421,417,412,415,417,413,410,419,416])
data2 =  np.array([415,420,413,409,423,419,411])

# Calculate basic statistics data 1
mean_val = np.mean(data1)
median_val = np.median(data1)
std_dev = np.std(data1, ddof=1) # Standard deviation
variance = np.var(data1, ddof=1) # Variance
min_val = np.min(data1) # Minimum value
max_val = np.max(data1) # Maximum value
n_count = len(data1)

# Calculate basic statistics data 2
mean_val2 = np.mean(data2)
median_val2 = np.median(data2)
std_dev2 = np.std(data2, ddof=1) # Standard deviation
variance2 = np.var(data2, ddof=1) # Variance
min_val2 = np.min(data2) # Minimum value
max_val2 = np.max(data2) # Maximum value
n_count2 = len(data2)

print(f"Mean1: {mean_val}, Mean2: {mean_val2}")
print(f"Median1: {median_val}, Median2: {median_val2}")
print(f"Standard Deviation1: {std_dev}, Standard Deviation2: {std_dev2}")
print(f"Variance1: {variance}, Variance2: {variance2}")
print(f"Minimum1: {min_val}, Minimum2: {min_val2}")
print(f"Maximum1: {max_val},Maximum2: {max_val2}")
print(f"n1: {n_count}, n2: {n_count2}")
```

```python
# t studen Distribition
probability = 0.025  # The accumulated probability of alpha/2 (e.g., for a 95% confidence interval, this is for one tail)
df = n_count - 1  # Degrees of freedom
# Get the t-value (critical value)
t_value = stats.t.ppf(probability, df)
print(t_value)
```

```python
import math
# Calculate the IC for the mean and varianze unknow using: X +/- t_(1-alpha/2,(n-1))*(stn_dev / (n)^0.5)
sr=math.sqrt(n_count)

IC_2=mean_val -(t_value*std_dev/sr)
IC_1=mean_val +(t_value*std_dev/sr)
print(IC_1, IC_2)
```

```python
# PREGUNTA 4

# (limite inferior IC)

# Define the degrees of freedom (df) and percentile
df1 = n_count-1  # Example: Degrees of freedom
percentile1 = 1-0.025  # Example: 99th percentile (area to the left)

# Calculate the chi-square value (critical value)
chi_square_value1 = stats.chi2.ppf(q=percentile1, df=df1)

print(f"The chi-square value for {df1} degrees of freedom and the {percentile1*100}th percentile is: {chi_square_value1}")

# (limite Superior IC)

# Define the degrees of freedom (df) and percentile
df2 = n_count-1  # Example: Degrees of freedom
percentile2 = 0.025  # Example: 95th percentile (area to the left)

# Calculate the chi-square value (critical value)
chi_square_value2 = stats.chi2.ppf(q=percentile2, df=df2)

print(f"The chi-square value for {df2} degrees of freedom and the {percentile2*100}th percentile is: {chi_square_value2}")
```

```python
IC_1 = ((n_count-1)*variance)/chi_square_value1
IC_2 = ((n_count-1)*variance)/chi_square_value2
print(IC_1,IC_2)
```

```python
# Define your parameters:
probability = 0.9  # The cumulative probability (e.g., for a 95% confidence level)
dfn = n_count2-1             # Numerator degrees of freedom (df1)
dfd = n_count-1            # Denominator degrees of freedom (df2)

# Calculate the F value (critical value)
f_value_1 = stats.f.ppf(q=probability, dfn=dfn, dfd=dfd)

probability2 = 0.1  # The cumulative probability (e.g., for a 95% confidence level)
dfn2 = n_count2-1             # Numerator degrees of freedom (df1)
dfd2 = n_count-1           # Denominator degrees of freedom (df2)

# Calculate the F value (critical value)
f_value_2 = stats.f.ppf(q=probability2, dfn=dfn2, dfd=dfd2)


print(f"The F value for a cumulative probability of {probability} is: {f_value_1}")
print(f"The F value for a cumulative probability of {probability2} is: {f_value_2}")
```

```python
IC_2f = variance/variance2*f_value_1
IC_1f = variance/variance2*f_value_2
print(IC_1f,IC_2f)
```
