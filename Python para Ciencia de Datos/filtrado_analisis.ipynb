```python
import numpy as np
import pandas as pd
# CSV
path = '/content/Online_Retail.csv'
sales_data = pd.read_csv(path, encoding='latin1')

# Convertir la columna "InvoiceDate" a tipo date Time
sales_data['InvoiceDate'] = pd.to_datetime(sales_data['InvoiceDate'])

#Eliminar las filas con valores faltantes en las columnas criticas
sales_data.dropna(subset=["CustomerID", "InvoiceDate"], inplace = True)

# Crear una nueva columna "TotalPrice"
sales_data["TotalPrice"] = sales_data["Quantity"] * sales_data["UnitPrice"]

print(sales_data.head())
```

```python
# Filtrar Ventas en el Reino Unido
UK_sales = sales_data[sales_data["Country"] == "United Kingdom"]
print(UK_sales)
```

```python
# Filtra pedidos mayores a 10 unidades
high_quantity_sales = sales_data[sales_data["Quantity"] > 10]
print(high_quantity_sales)
```

```python
# Filtrar Ventas en el Reino Unido y cantidades mayores a 10 nombre_data[(nombre_data["No_encabezado_colum"] == "Nombre_variable_interes") & (nombre_data["No_encabezado_colum_2"] > la condicion)]
UK_high_quantity = sales_data[(sales_data["Country"] == "United Kingdom") & (sales_data["Quantity"] > 10)]
print(UK_high_quantity)
```

```python
# Filtracion de eventas en el año 2011, nombre_data[nombre_data["No_encabezado_colum"].dt.year == año a comparar, .dt.year extrae el año

sales_2011 = sales_data[sales_data["InvoiceDate"].dt.year == 2011]
print(sales_2011)
```

```python
sales_dec_2010 = sales_data[(sales_data["InvoiceDate"].dt.year == 2010) & (sales_data["InvoiceDate"].dt.month == 12)]
print(sales_dec_2010)
```

```python
pivot_table = pd.pivot_table(sales_data, values ="Quantity", index = "Country", columns="StockCode", aggfunc="sum")
print(pivot_table)
```
