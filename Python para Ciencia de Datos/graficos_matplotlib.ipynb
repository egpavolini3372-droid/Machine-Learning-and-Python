```python
import numpy as np
import matplotlib.pyplot as plt

month = np.array(["Enero","Febrero","Marzo","Abril","Mayo"])
sales = np.array([20,25,30,28,35])

# Configurar el tamano del grafico
plt.figure(figsize=(8,6))

# crear el grafico
plt.plot(month, sales, marker = 'o',color = "blue")
plt.title("ventas mensuales dle producto")
plt.xlabel("meses")
plt.ylabel("ventas")
plt.grid(True)

# Mostrar el grafico
plt.show()
```

```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt

# Configurar el tamano del grafico
plt.figure(figsize=(8,6))

hours = [2,3,4,5,6,7,8]
result_1 = [55,60,65,70,75,80,85]
result_2 = [50,58,63,74,75,78,83]

# Crear Gráfico de dispersión de dos estudiantes

plt.scatter(hours,result_1, marker = "o", color ="green", label = "Estudiante 1")
plt.scatter(hours,result_2, marker = "s", color ="red", label = "Estudiante 2")


plt.title("Relacion horas estudiadas y performance")
plt.xlabel("horas estudiadas")
plt.ylabel("Performance")
plt.grid(True)
# Mostrar el grafico
plt.show()
```

```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt

# Configurar el tamano del grafico
plt.figure(figsize=(8,6))

hours = [2,3,4,5,6,7,8]
result_1 = [55,60,65,70,75,80,85]
result_2 = [50,58,63,69,74,78,83]

# Crear Gráfico de dispersión de dos estudiantes

plt.plot(hours,result_1, marker = "o", color ="green",linestyle='-',linewidth=2, label = "Estudiante 1")
plt.plot(hours,result_2, marker = "s", color ="red",linestyle="--",linewidth=2, label = "Estudiante 2")


plt.title("Relacion horas estudiadas y performance")
plt.xlabel("horas estudiadas")
plt.ylabel("Performance")
plt.grid(True)
# Mostrar el grafico
plt.show()
```
